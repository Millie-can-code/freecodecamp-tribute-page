      <div id="img-caption">🧠 <b>Who Was George Washington Carver?</b>
        <p>George Washington Carver (c. 1864–1943) was an American agricultural scientist, inventor, and educator.
        He was born into slavery but went on to become one of the most respected scientists of his time, dedicated to improving the lives of poor farmers in the southern United States.
        </p>
      </div>
    </div>
    <section id="tribute-info">
      <h2>🌾 What He Was Popularly Known For</h2>
      <fieldset>
      <h3>1. Promoting Crop Rotation</h3>
      <ul>
        <li>Carver encouraged farmers to <b>rotate crops</b> (like planting peanuts, soybeans, and sweet potatoes instead of cotton every year).</li>
        <li>This restored <b>nitrogen to the soil</b>, which cotton farming had depleted.</li>
        <li>It helped <b>revitalize farmlands</b> across the American South and improved crop yields.</li>
      </ul>
      <p>🧩 Why it mattered: His work stopped soil exhaustion and gave farmers more sustainable income sources.</p>
      </fieldset>
      <fieldset>
      <h3>2. Inventing Hundreds of Products</h3>
      <ul>
        <li>Carver developed over <b>300 products</b> from peanuts, including:</li>
        <ul><li>Dyes</li>
        <li>Paints</li>
        <li>Soap</li>
        <li>Cooking Oli</li>
        <li>Cosmetics</li>
        <li>Plastic substitutes</li>
        <li>Paper</li>
        <li>Fuel</li></ul>
        <li>He also made over <b>100 products from sweet potatoes</b>, like flour, vinegar, and glue.</li>
      </ul>
      <p>🥜 Fun fact: Contrary to myth, Carver didn’t invent peanut butter — but he made it</p>
      </fieldset>
      <fieldset>
      <h3>3. Helping Poor Farmers</h3>
      <ul>
        <li>Carver’s main goal wasn’t profit — it was <b>education and empowerment</b>.</li>
        <li><b>He also taught farmers how to</b>:</li>
        <ul><li>Improve soil with crop rotation</li>
        <li>Use natural fertilizers</li>
        <li>Make homemade goods to sell and use</li></ul>
      </ul>
      <p>💡 He even created a mobile classroom, called the
       <b>“Jesup Wagon”</b>, to teach rural farmers practical agricultural techniques.</p>
      </fieldset>
      <fieldset>
      <h3>4. Work at Tuskegee Institute</h3>
      <ul>
        <li>Carver spent over <b>40 years teaching and researching</b> at <b>Tuskegee Institute</b> (now Tuskegee University) in Alabama.</li>
        <li>He turned the small agricultural department into a respected research center.</li>
        <li>He emphasized <b>self-reliance, environmental care, and education</b>.</li>
      </fieldset>
      <fieldset>
      <h3>5. Recognition and Legacy</h3>
      <ul>
        <li>Known as the <b>“Plant Doctor”</b> since childhood.</li>
        <li>He was an <b>advisor to U.S. presidents</b> (like Theodore Roosevelt and Franklin D. Roosevelt).</li>
        <li>Received honors such as:</li>
        <ul><li><b>Spingarn Medal</b> from the NAACP (1923)</li>
        <li><b>Carver National Monument</b> — the first U.S. national monument dedicated to a Black American.</li></ul>
        <p>🌟 Legacy: His work in agriculture, sustainability, and education continues to influence scientists and environmentalists worldwide.</p>
      </fieldset>
      <fieldset>
      <h3>✨ Famous Quote</h3>
        <p>“It is not the style of clothes one wears, neither the kind of automobile one drives, nor the amount of money one has in the bank, that counts. These mean nothing. It is simply service that measures success.”</p>
      </fieldset>
